```python
def greet(name: str) -> str:
    """produced Hello, name!"""
    return "Hello, " + name + "!"

def test_greet():
    assert greet("Alice") == "Hello, Alice!"
    assert greet("Bob") == "Hello, Bob!"
```

```python
def statement(name: str, age: int) -> str:
    """returns a string"""
    return name + " is " + age + " years old."

def test_statement():
    assert statement("Alice", 20) == "Alice is 20 years old."
```

```python
def shipping_cost(weight: float) -> int:
    """if weight <= 1, cost is 5; if weight <= 5, cost is 10; otherwise 20"""
    if weight <= 1:
        return 5
    elif weight <= 5:
        return 10
    else:
        return 20

def test_shipping_cost():
    assert shipping_cost(0.2) == 5
    assert shipping_cost(2.0) == 10
    assert shipping_cost(10.0) == 20
    
```

```python
def grade_letter(score: int) -> str:
    """returns "A" for scores 90 and above, "B" for 80–89, "C" for 70–79, "D" for 60–69, and "F" otherwise."""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        "F"

def test_grade_letter():
    grade_letter(90) == "A"
    grade_letter(82) == "B"
    grade_letter(75) == "C"
    grade_letter(60) == "D"
    grade_letter(12) == "F"
    
```

```python
class Book:
    title: str
    author: str
    pages: int

books = [
    Book("To Kill a Mockingbird", "Harper Lee", 285),
    Book("1984", "George Orwell", 328),
    Book("Pride and Prejudice", "Jane Austen", 432),
    Book("The Great Gatsby", "F. Scott Fitzgerald", 180)
]

long_books = list(filter(lambda books: books.pages > 300, books))
print(long_books)
```
